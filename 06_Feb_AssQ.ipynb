```python
# Q1. Create a function which will take a list as an argument and return the product of all the numbers
# after creating a flat list.
# Use the below-given list as an argument for your function.
# list1 = [1,2,3,4, [44,55,66, True], False, (34,56,78,89,34), {1,2,3,3,2,1}, {1:34, "key2": [55, 67, 78, 89], 4: (45,
# 22, 61, 34)}, [56, 'data science'], 'Machine Learning']
# Note: you must extract numeric keys and values of the dictionary also.
```

```python
        
def product_of_list(list1):
    flatten_list = flat_list(list1)
    product = 1 

    for i in flatten_list:
        product *= i
    return product

def flat_list(list1):
    list2 = []
    for data in list1:
        if isinstance(data, (int, float, bool)):
            list2.append(data)
        elif isinstance(data, (list, tuple, set)):
            list2.extend(flat_list(data))
        elif isinstance(data, dict):
            for key, value in data.items():
                if isinstance(key, (int, float, bool)):
                    list2.append(key)
                if isinstance(value, (int, float, bool)):
                    list2.append(value)
                elif isinstance(value, (list, tuple, set)):
                    list2.extend(flat_list(value))
                
    return list2

```

```python
# Q2. Write a python program for encrypting a message sent to you by your friend. The logic of encryption
# should be such that, for a the output should be z. For b, the output should be y. For c, the output should
# be x respectively. Also, the whitespace should be replaced with a dollar sign. Keep the punctuation
# marks unchanged.
# Input Sentence: I want to become a Data Scientist.
# Encrypt the above input sentence using the program you just created.
# Note: Convert the given input sentence into lowercase before encrypting. The final output should be
# lowercase.

i$wznt$yexome$zdztz$sxientist
```

```python
def encryption_msg(message):
    msg = ""
    for i in message.lower():
        if i == "a":
           msg += "z"
        elif i == "b":
            msg += "y"
        elif i == "c":
            msg += "x"
        elif i == " ":
            msg += "$"
        else:
            msg += i
    return msg
        
encryption_msg("I want to become a Data Scientist")
```

```python
print("I want to become a Data Scientist.".lower().replace(" ", "$").replace("a", "z").replace("b", "y").replace("c", "x"))
```
