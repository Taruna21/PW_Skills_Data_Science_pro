```python
#Q2. Given a Pandas DataFrame df with columns 'A', 'B', and 'C', write a Python function to re-index the
#DataFrame with a new index that starts from 1 and increments by 2 for each row.

import pandas as pd

column = ["A", "B", "C"]
rows = [["1", "2", "3"],
        ["4", "5", "6"],
        ["7", "8", "9"]]

df = pd.DataFrame(rows, columns=column)
print(df)
df.index = range(1, len(df)*2,2)
print(df)
```

```python
import pandas as pd

column = ["A", "B", "C"]
rows = [["1", "2", "3"],
        ["4", "5", "6"],
        ["7", "8", "9"]]

# creating DataFrame
df = pd.DataFrame(rows, columns=column)

# changing index
df.index = range(1, len(df) * 2, 2)

print(df)
```

```python
"""Q4. Given a Pandas DataFrame df with a column 'Text', write a Python function to create a new column
'Word_Count' that contains the number of words in each row of the 'Text' column."""

Text = ["I am Taruna", "Hello World", "Pandas is easy"]
df = pd.DataFrame(Text, columns=["Text"])
print(df)
df["Word_count"] = df["Text"].apply(lambda x:len(x.split()))
print(df)
```

```python
"""Q7. You have a Pandas DataFrame df that contains a column named 'Email' that contains email
addresses in the format 'username@domain.com'. Write a Python function that creates a new column
'Username' in df that contains only the username part of each email address."""


Email = ["taruna@domain.com", "d@domain.com"]

df = pd.DataFrame(Email, columns=["Email"])
print(df)
df["Username"] = df["Email"].apply(lambda x: x.split("@")[0])

print(df)
```

```python
"""Q10. Given a Pandas DataFrame df with a column 'Sales' and a column 'Date', write a Python function to
create a new column 'MovingAverage' that contains the moving average of the sales for the past 7 days
for each row in the DataFrame. The moving average should be calculated using a window of size 7 and
should include the current day."""


Sales = [1,2,3,4,5,6,7]
Date = ['2026-05-06', '2026-05-07', '2026-05-08', '2026-05-09', '2026-05-10', '2026-05-11', '2026-05-12']

df = pd.DataFrame({"Sales": Sales, "Date": Date})
print(df)
```

```python
df["MovingAverage"] = df["Sales"].rolling(window=7).mean()

print(df)
```

```python
import pandas as pd

Date = ['2023-01-05', '2023-01-15', '2023-02-10',
        '2023-01-25', '2023-03-01']

df = pd.DataFrame({"Date": Date})

# converting Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# selecting rows between two dates
filtered_df = df[
    (df["Date"] >= "2023-01-01") &
    (df["Date"] <= "2023-01-31")
]

print(filtered_df)
```
